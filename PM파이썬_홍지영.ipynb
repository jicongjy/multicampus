```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                    # 처리: 부가세 10%를 더한 금액 계산
print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
```

```python
onion = int(input("양파 가격(원): "))
tomato = int(input("토마토 가격(원): "))
potato = int(input("감자 가격(원): "))

print("양파 가격:", onion*total, "원")
print("토마토 가격:", tomato*total, "원")
print("감자 가격:", potato*total, "원")
```

```python
onion_base = int(input("양파 기본 가격(원): "))
tomato_base = int(input("토마토 기본 가격(원): "))
potato_base = int(input("감자 기본 가격(원): "))

tax_rate = 1.1
onion_total = int(onion_base * tax_rate)
tomato_total = int(tomato_base * tax_rate)
potato_total = int(potato_base * tax_rate)

print(f"양파 결제 금액: {onion_total}원")
print(f"토마토 결제 금액: {tomato_total}원")
print(f"감자 결제 금액: {potato_total}원")
```

```python
# 1. 이전 메뉴팜
print("="*30)
print("     [ 메뉴판 ]")
print("="*30)
print(f"양파 단가: {onion_total}원")
print(f"토마토 단가: {tomato_total}원")
print(f"감자 단가: {potato_total}원")
print("="*30)
print()

# 2. 입력
onion_num = int(input("양파 몇 개 사시겠어요? "))
tomato_num = int(input("토마토 몇 개 사시겠어요? "))
potato_num = int(input("감자 몇 개 사시겠어요? "))

# 3. 계산
onion_sum = onion_total * onion_num
tomato_sum = tomato_total * tomato_num
potato_sum = potato_total * potato_num
total_sum = onion_sum + tomato_sum + potato_sum

# 4. 출력
print("\n" + "="*30)
print("          [ 주문 영수증 ]")
print("="*30)
print(f"양파: {onion_num}개 = {onion_sum}원")
print(f"토마토: {tomato_num}개 = {tomato_sum}원")
print(f"감자: {potato_num}개 = {potato_sum}원")
print("-"*30)
print(f"총 결제 금액은 {total_sum}원 입니다.")
print("="*30)
```
